# research question: Are certain terrain characteristics associated with more severe avalanche accidents?

```python
# copy data frame
df_terrain = df.loc[:, [
    "avalanche_id",
    "date",
    "canton",
    "elevation",
    "slope_aspect",
    "inclination",
    "dangerlevel_rating1",
    "dangerlevel_rating1_subdivision",
    "dangerlevel_rating2",
    "deaths",
    "caught",
    "fully_buried",
    "activity"
]].copy()
df_terrain.head()
```

```python
# missing values for the relevant variables
df_terrain.isna().sum()
```

```python
# show missing values as a percentage
missing_percent = (df_terrain.isna().mean() * 100)
print(missing_percent)
```

```python
# fill in missing values for inclination
df_terrain["inclination"] = (df_terrain["inclination"].fillna("Unknown"))
```

```python
# test for duplicates
print(df_terrain.duplicated().sum())
print(df_terrain["avalanche_id"].duplicated().sum())
```

# Cleaning phase

```python
# check types
df_terrain.dtypes
```

```python
# to datetime 
df_terrain["date"] = pd.to_datetime(df_terrain["date"], errors="coerce")
df_terrain.head()
```

```python
# to category
df_terrain["inclination"].value_counts(dropna=False)
df_terrain["inclination"] = (df_terrain["inclination"].astype("category"))
```

```python
# to category
df_terrain["slope_aspect"].value_counts(dropna=False)
df_terrain["slope_aspect"] = (df_terrain["slope_aspect"].astype("category"))
```
