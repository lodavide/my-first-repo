```python
from pathlib import Path

path = Path("practice_input.txt")
text2 = path.read_text(encoding="utf-8")

clean_lines = []

for line in text2.splitlines():
    line = line.strip()

    if line:
        clean_lines.append(line.upper())

clean_text = "\n".join(clean_lines)

Path("practice_clean.txt").write_text(clean_text, encoding="utf-8")
print(clean_text)
```

For a non-negative integer duration, return `(hours, remaining_minutes)`. Unpack one result into two variables.

### 8.4

Return `(total, average)` for a list of numbers. For an empty list, return `(0, None)`.


```python
def summarise(values):
    total = sum(values)

    if len(values) == 0:
        return total, None
    
    average = total/ len(values)
    return total, average

print(summarise([2, 0, 4]))
print(summarise([]))

```

### 8.6

Complete the helpers and dispatcher:

- `difference(a, b)`: return `a - b`.
- `percentage(a, b)`: return `a / b * 100`, or None if `b == 0`.
- `calculate(a, b, operation)`: call the matching helper for `"difference"` or `"percentage"`; return None for other operations or non-numeric inputs. Assume booleans are not supplied.


```python
def difference(a, b):
    return a - b


def percentage(a, b):
    if b == 0:
        return None
    return (a / b) * 100


def calculate(a, b, operation):
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        return None
    if operation == "difference":
        return difference(a, b)
    elif operation == "percentage":
        return percentage(a, b)
    else:
        return None
    



print(calculate(15, 20, 'difference'))
print(calculate(15, 20, 'percentage'))
print(calculate(1, 0, 'percentage'))
print(calculate('1', 2, 'difference'))
print(calculate(1, 2, 'other'))
```
